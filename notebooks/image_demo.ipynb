### What this notebook covers

1. Generating a synthetic RGB gradient image
2. Converting to grayscale
3. Resizing and rotating
4. Computing pixel statistics with NumPy


## 1. Setup

Import the required libraries and configure the output directory.

```python
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
import numpy as np
import os

OUTDIR = os.path.join('..', 'images')
os.makedirs(OUTDIR, exist_ok=True)

print('Output directory:', os.path.abspath(OUTDIR))
```

## 2. Create a Gradient Image

Build a 400x300 image where red increases horizontally and green increases vertically.

```python
width, height = 400, 300
img = Image.new('RGB', (width, height))

for y in range(height):
    for x in range(width):
        r = int(255 * x / width)
        g = int(255 * y / height)
        b = 128
        img.putpixel((x, y), (r, g, b))

path = os.path.join(OUTDIR, 'sample_image.png')
img.save(path)
print('Saved:', path)
print('Size:', img.size)
```

## 3. Display the Original Image

```python
plt.figure(figsize=(8, 6))
plt.imshow(img)
plt.title('Original Color Image')
plt.axis('off')
plt.show()
```

## 4. Grayscale Conversion

```python
gray = ImageOps.grayscale(img)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(img)
axes[0].set_title('Original')
axes[0].axis('off')
axes[1].imshow(gray, cmap='gray')
axes[1].set_title('Grayscale')
axes[1].axis('off')
plt.tight_layout()
plt.show()

path = os.path.join(OUTDIR, 'sample_gray.png')
gray.save(path)
print('Saved:', path)
```

## 5. Resize and Rotate

```python
resized = img.resize((200, 150))
rotated = img.rotate(45, expand=True)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img)
axes[0].set_title('Original (400x300)')
axes[0].axis('off')
axes[1].imshow(resized)
axes[1].set_title('Resized (200x150)')
axes[1].axis('off')
axes[2].imshow(rotated)
axes[2].set_title('Rotated 45 deg')
axes[2].axis('off')
plt.tight_layout()
plt.show()

resized.save(os.path.join(OUTDIR, 'sample_resized.png'))
rotated.save(os.path.join(OUTDIR, 'sample_rotated.png'))
print('Saved resized and rotated images')
```

## 6. Pixel Statistics

```python
arr = np.array(img)
print('Image shape:', arr.shape)
print('Min pixel value:', arr.min())
print('Max pixel value:', arr.max())
print('Mean pixel value:', round(float(arr.mean()), 2))
```
